```python

# =====================================================================
# STEP 3: INTERACTIVE DECODING ENGINE & LOCAL CSV DATABASE LOGGER
# =====================================================================
import csv
from datetime import datetime
import os

class ColabIndusEngine:
    def __init__(self, db_filename="indus_seals_database.csv"):
        self.db_filename = db_filename
        self.consonants = {
            '🏺': 'Ka', '𓏎': 'Kha', '𓏍': 'Ga', '𓎏': 'Ca', '𓎐': 'Cha',
            '𓏲': 'Ta', '🏹': 'Ta', '𓊍': 'Dha', '𓏏': 'Dha', '𓍢': 'Na',
            '🧍': 'Pa', '𓀜': 'Pha', '𓆛': 'Ma', '◯': 'Ra', '☉': 'Ra',
            '🌿': 'Sha', '𓇚': 'Sha'
        }
        # Initialize file structure with master tracking headers if file doesn't exist
        if not os.path.exists(self.db_filename):
            with open(self.db_filename, mode='w', newline='', encoding='utf-8') as f:
                writer = csv.writer(f)
                writer.writerow(["Timestamp", "Seal_Identifier", "Sign_Sequence", "Decoded_Sanskrit", "Prashasanik_Meaning"])

    def decode_seal(self, seal_id, sequence_input):
        tokens = [t.strip() for t in sequence_input.split('+')]
        decoded_phonetics = []

        # Check Alpha Prefix 'Mahā'
        if tokens and (tokens[0] == '𓎦' or tokens[0] == '37'):
            decoded_phonetics.append("Mahā")
            tokens = tokens[1:]

        for i, token in enumerate(tokens):
            if i == len(tokens) - 1: # Terminal Inflection Check
                if token in ['𓆛', '99']: decoded_phonetics.append("m"); continue
                elif token in ['𓆜', '162']: decoded_phonetics.append("asya"); continue
                elif token in ['𓆟', '99_wing']: decoded_phonetics.append("asti"); continue

            # Posture Mutations & Structural Ligatures
            if token in ['𓀠', '311_R'] and '⚙' in tokens:
                decoded_phonetics.append("Bhārata")
                if '⚙' in tokens: tokens.remove('⚙')
            elif token in ['⚙', '245_S']:
                if i + 1 < len(tokens) and tokens[i+1] in ['⚙', '245_S']:
                    decoded_phonetics.append("Rājarāja")
                    tokens.remove(tokens[i+1])
                else:
                    decoded_phonetics.append("Rāja")
            elif token in ['🌿', '𓇚'] and i + 1 < len(tokens) and tokens[i+1] in ['𓀜', '311_W']:
                decoded_phonetics.append("Śubha"); tokens.remove(tokens[i+1])
            elif token in ['𓍢', 'hook'] and i + 1 < len(tokens) and tokens[i+1] in ['𓀜', '311_W']:
                decoded_phonetics.append("Lābha"); tokens.remove(tokens[i+1])
            elif token in self.consonants:
                decoded_phonetics.append(self.consonants[token])
            else:
                decoded_phonetics.append(token)

        raw_sanskrit = "-".join(decoded_phonetics).replace("-m", "m").replace("-asya", "asya").replace("-asti", "asti")
        meaning = "Sovereign State Proclamation"
        if "Bhārata" in raw_sanskrit: meaning = "Imperial Jurisdiction Pass (Central Bharata Grid)"
        elif "Rājarāja" in raw_sanskrit or "Mahā" in raw_sanskrit: meaning = "Sovereign Audit Token from Maharaja Vaults"
        elif "Śubha" in raw_sanskrit or "Lābha" in raw_sanskrit: meaning = "Civic Gateway Maritime Customs Document"

        # Lock entry to database file
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        with open(self.db_filename, mode='a', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow([timestamp, seal_id, sequence_input, raw_sanskrit, meaning])

        return raw_sanskrit, meaning

# Trigger Core Loop Execution inside Colab
engine = ColabIndusEngine()
print("=== सिंधु-पाणिनीय-सान्दीपनि क्लाउड डिकोडिंग इंजन ===")
while True:
    seal_num = input("\n१. सील संख्या (Type 'exit' to finalize and push): ").strip()
    if seal_num.lower() == 'exit':
        print("✓ Translation loop finished. Ready for automated cloud push.")
        break
    sign_chain = input("२. संकेत अनुक्रम (e.g. 𓎦+311_R+⚙+𓆜): ").strip()
    ans, mean = engine.decode_seal(seal_num, sign_chain)
    print(f"-> डिकोडेड संस्कृत: {ans}\n-> प्रशासनिक अर्थ: {mean}")
```
